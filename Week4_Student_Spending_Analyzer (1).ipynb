##Section 2: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

* **Pandas:** Used to create, organize, explore, and group the dataset.
* **NumPy:** Used to calculate numerical statistics.
* **Matplotlib:** Used to create data visualizations like bar charts.

##Section 3: Create the Dataset

```python
data = {
    "Category": [
        "Food", "Transport", "Education", "Food",
        "Entertainment", "Transport", "Food",
        "Education", "Entertainment", "Transport",
        "Food", "Education"
    ],
    "Amount": [
        500, 150,
        800, 350,
        400, 200,
        600,
        700, 250,
        180,
        450, 900
    ],
    "Payment_Method": [
        "Cash", "Card",
        "Online", "Cash",
        "Card", "Cash",
        "Online",
        "Online", "Card", "Cash",
        "Cash", "Online"
    ]
}

df = pd.DataFrame(data)
display(df)
```

The dataset contains three columns:
* **Category:** The type of expense.
* **Amount:** The expense amount in arbitrary units.
* **Payment_Method:** How the expense was paid.

# Section 4: Explore the Data
Using Pandas to inspect the dataset's rows, columns, and summary information.

```python
df.head()
df.shape
df.info()
df.describe(include="all")
```

The `.shape` command shows the dataset contains 12 rows and 3 columns. The `.info()` and `.describe()` outputs confirm the data types and provide a high-level summary of the entries.

# Section 5: Check Data Quality

```python
df.isnull().sum()
df.duplicated().sum()
```

The output shows that there are 0 missing values and 0 duplicate rows, meaning the data is clean and ready for analysis.

# Section 6: Calculate Statistics

```python
# Question 1: What is the total spending across all entries?
total_spending = df["Amount"].sum()
print("Total spending:", total_spending)

# Question 2: What is the average spending per entry?
average_spending = np.mean(df["Amount"])
print("Average spending:", average_spending)

# Question 3: Which category has the highest total spending?
category_spending = df.groupby("Category")["Amount"].sum()
print("\nSpending by Category:\n", category_spending)
highest_category = category_spending.idxmax()
print("\nCategory with the highest spending:", highest_category)

# Question 4: What is the highest individual expense?
highest_expense = df["Amount"].max()
print("\nHighest individual expense:", highest_expense)

# Question 5: How much was spent through each payment method?
payment_spending = df.groupby("Payment_Method")["Amount"].sum()
print("\nSpending by Payment Method:\n", payment_spending)
```

The total spending is 5480 units, with an average of about 456.67 units per transaction. Education was the highest overall spending category (2400 units), and the largest single expense was also 900 units. Online payments accounted for the majority of the total spending.

# Section 7: Create a Bar Chart
Visualizing total spending by category to identify major expenses.

```python
category_spending = df.groupby("Category")["Amount"].sum()
category_spending.plot(kind="bar")
plt.title("Total Spending by Category")
plt.xlabel("Expense Category")
plt.ylabel("Total Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

The visualization clearly shows that the "Education" category has the highest total spending. The bar chart makes it easy to instantly compare categories without having to read through the raw numerical data.
